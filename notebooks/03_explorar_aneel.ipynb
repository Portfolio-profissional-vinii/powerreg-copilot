```python
import pandas as pd
from pathlib import Path

# Localiza o arquivo CSV dentro da pasta data/raw/aneel/
data_path = Path("../data/raw/aneel")
csv_files = list(data_path.glob("*.csv"))

if csv_files:
    arquivo_csv = csv_files[0]
    print(f"📁 Arquivo localizado: {arquivo_csv.name}\n")
    
    # Leitura inicial das primeiras 5 linhas
    # (Dados da ANEEL costumam usar separador ';' e encoding 'utf-8' ou 'latin1')
    try:
        df = pd.read_csv(arquivo_csv, sep=";", encoding="utf-8", nrows=5)
    except UnicodeDecodeError:
        df = pd.read_csv(arquivo_csv, sep=";", encoding="latin1", nrows=5)
        
    print("📋 Colunas encontradas:")
    for col in df.columns:
        print(f" - {col}")
        
    print("\n🔍 Amostra dos Dados:")
    display(df.head())
else:
    print("⚠️ Nenhum arquivo CSV foi encontrado na pasta 'data/raw/aneel/'. Certifique-se de colar o arquivo lá!")
```
